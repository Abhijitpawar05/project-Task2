```python
#Student Performance Distribution Analysis

This project focuses on analyzing student performance using statistical plots and distribution analysis.
The dataset contains information about students’ study hours, sections, and marks.
Exploratory Data Analysis (EDA) is performed to understand the distribution of marks and identify patterns in the data.

Histograms and KDE plots are used to visualize the overall distribution of marks, while boxplots help in identifying the spread, median, and outliers.
Section-wise comparison is also performed to compare the performance of students from different sections.

The analysis helps in understanding skewness, variability, and performance trends among students.
This project demonstrates how statistical visualizations can be effectively used to gain insights from data before applying advanced analytical techniques.
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
## *** Histrogram
plt.hist(df['Marks'], bins=10)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Histogram of Student Marks", fontweight='bold')
plt.show()
```

```python
## KDE Plot
sns.kdeplot(df['Marks'], fill=True)
plt.title("KDE Plot of Student Marks")
plt.show()
```

```python
#Box Plot
sns.boxplot(y=df['Marks'])
plt.title("Boxplot of Student Marks")
plt.show()
```

```python
##Group of Comparison (A vs B)
sns.boxplot(x='Section', y='Marks', data=df)
plt.title("Marks Comparison by Section")
plt.show()
```

```python
#Skewness
df['Marks'].skew()
```
